*Recordatorio breve de tipos de variables*

**Cuantitativa discreta**
<br>
Información basada en conteos. Admite un número finito de valores y esos valores no se pueden subdividir de forma significativa.

**Cualitativa nominal**
<br>
Variables sin orden ni secuencia de clasificación inherente. Son etiquetas: nombran a qué grupo pertenece cada caso sin que ningún grupo sea más que otro.

**Cualitativa ordinal**
<br>
Variables cuyas categorías forman una serie ordenada. Se puede decir que una categoría está por encima de otra pero no cuánto exactamente, es por ello que la distancia entre nieveles no es medible ni necesariamente constante.

**Cualitativa binaria**
<br> Variables con solo dos opciones posibles (si/no, aprobado/reprobado). En rigor, es un caso particular de las anteriores ya que una binaria puede ser nominal o ordinal según si sus dos valores tienen orden pero podriamos tratarla como categoría aparte porque en la práctica este tipo de variables recibe un tratamiento distinto al codificarla (mapeamos directaemnte a 0/1 en vez de one hot encoding que menciono Maira)

2. Identifica con la descripción de los datos si podrían existir posibles datos atípicos.

Primero realizamos listas de para separar los Crímenes de los Derechos Humanos según su tipo de puntuación. Esto pues la clasificación en cada criterio es distinto y tiene un significado distinto.

```python
# Listas con los nombres de las columnas de crimenes y derechos ordenados por tipo de puntuación
# Columnas de Derechos con puntuaciones en 0 o 1
Rights_0_to_1 = [
    "Freedom of Movement (Old)",
    "Freedom of Religion (Old)"
]

# Columnas de Crímenes con puntuaciones en 0, 1 o 2
Crimes_0_to_2 = [
    "Disappearance",
    "Extrajudicial Killing",
    "Political Imprisonment",
    "Torture"
]

# Columnas de Derechos Humanos con puntuaciones en 0, 1 o 2
Rights_0_to_2 = [
    "Freedom of Assembly and Association",
    "Freedom of Foreign Movement",
    "Freedom of Domestic Movement",
    "Freedom of Speech",
    "Electoral Self-Determination",
    "Freedom of Religion (New)",
    "Employment Rights",
    "Independence of Judiciary",
]

# Columnas de Derechos Humanos con puntuaciones en 0, 1, 2 o 3
Rights_0_to_3 = [
    "Women’s Economic Rights",
    "Women’s Political Rights",
    "Women’s Social Rights"
]
```

Luego de pensarlo, se optó por un análisis de frecuencias relativas (porcentajes) por variable, pues para datos categóricos, un valor se considera un "outlier por baja frecuencia" si su tasa de aparición es por debajo de un umbra outlier (en este caso, 10% de los datos de esa columna).

```python
# Umbral crítico (ej: si un valor aparece en menos del 10% de los casos, es outlier)
UMBRAL_OUTLIER = 0.10

print(f"\t\t\t OUTLIERS POR BAJA FRECUENCIA DE DERECHOS HUMANOS EN ESCALA 0 O 1")

# Iterar sobre cada derecho
for columna in df_clean[Rights_0_to_1]:
    # Calculo de frecuencia relativa (%) omitiendo NaN
    frecuencias_relativas = df_clean[columna].value_counts(normalize=True)
    # Buscar si alguno de los puntajes está por debajo del umbral
    outliers_detectados = frecuencias_relativas[frecuencias_relativas < UMBRAL_OUTLIER]
    if not outliers_detectados.empty:
        print(f"\nVariable: {columna}")
        for valor, porcentaje in outliers_detectados.items():
            print(
                f"\t- El valor {int(valor)} es un OUTLIER: Solo representa el {porcentaje*100:.2f}% de los datos. NAN: {(df_clean[columna].isna().mean())*100:.2f}%"
            )
    else:
        print(f"* Variable: {columna} -> No tiene outliers (datos bien distribuidos). NAN: {(df_clean[columna].isna().mean())*100:.2f}%")
```
